```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
categorical_cols = X_train.select_dtypes(
    include=['object', 'category']
).columns.tolist()

numerical_cols = X_train.select_dtypes(
    include=['number']
).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nNumber of categorical columns:", len(categorical_cols))

print("\nNumerical columns:")
print(numerical_cols)

print("\nNumber of numerical columns:", len(numerical_cols))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_cols
        ),
        (
            'numerical',
            'passthrough',
            numerical_cols
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

logistic_model.fit(X_train_processed, y_train)

print("Logistic Regression model trained successfully.")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_cols
        ),
        (
            'numerical',
            StandardScaler(),
            numerical_cols
        )
    ]
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

logistic_model.fit(X_train_processed, y_train)

print("Logistic Regression model trained successfully.")
```

```python
y_pred = logistic_model.predict(X_test_processed)

print("First 20 predictions:")
print(y_pred[:20])

print("\nPrediction counts:")
print(pd.Series(y_pred).value_counts())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

y_pred_proba = logistic_model.predict_proba(X_test_processed)[:, 1]
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))
```

```python
import pandas as pd
import joblib

# Load saved model and preprocessor
model = joblib.load('medical_no_show_model.pkl')
preprocessor = joblib.load('medical_no_show_preprocessor.pkl')

def predict_no_show(input_data):
    
    # Convert input into DataFrame
    input_df = pd.DataFrame([input_data])
    
    # Preprocess input
    input_processed = preprocessor.transform(input_df)
    
    # Prediction
    prediction = model.predict(input_processed)[0]
    
    # Probability of no-show
    probability = model.predict_proba(input_processed)[0][1]
    
    if prediction == 1:
        result = "No-show"
    else:
        result = "Attended"
    
    return result, probability


print("Prediction function created successfully.")
```

```python
def predict_appointment(input_data):
    
    input_df = pd.DataFrame([input_data])
    
    # Preprocess input
    input_processed = preprocessor.transform(input_df)
    
    # Get prediction probabilities
    probabilities = model.predict_proba(input_processed)[0]
    
    attended_probability = probabilities[0]
    no_show_probability = probabilities[1]
    
    # Final prediction
    prediction = "No-show" if no_show_probability >= 0.50 else "Attended"
    
    return {
        "Prediction": prediction,
        "Attended Probability": round(attended_probability * 100, 2),
        "No-show Probability": round(no_show_probability * 100, 2)
    }


# Test the function
prediction_result = predict_appointment(sample_appointment)

print(prediction_result)
```
